# SecuMail 2.0 : installation

Menu *Exécution* → « Tout exécuter », puis acceptez l'accès à votre Google Drive.
Le code est copié dans le dossier `secumail/` de votre Drive.

```python
import io
import urllib.request
import zipfile
from pathlib import Path

from google.colab import drive

URL = "https://github.com/HamzaEzzRa/Gentech-SecuMail/releases/latest/download/secumail.zip"

drive.mount("/content/drive")
mon_drive = Path("/content/drive/MyDrive")

if (mon_drive / "secumail").exists():
    print("Le dossier secumail/ existe déjà : votre travail n'a pas été modifié.")
else:
    with urllib.request.urlopen(URL) as reponse:
        zipfile.ZipFile(io.BytesIO(reponse.read())).extractall(mon_drive)
print("Installation terminée")
```
