## Le piège des petits nombres

`URGENT: RE: TP3` est tout en majuscules, et parfaitement légitime.

```python
texte = "URGENT: RE: TP3"
lettres = [c for c in texte if c.isalpha()]
majuscules = [c for c in lettres if c.isupper()]
print(len(lettres), "lettres,", len(majuscules), "majuscules")
```

# Un mot-clé ou une expression ?

Une expression exige que plusieurs mots apparaissent **ensemble, dans cet ordre**.

| Texte | Analyse | Points |
|---|---|---|
| « croissance de 12 % des ventes » | pourcentage seul | rien |
| « -50 % de rabais » | pourcentage + rabais | **+10** |
